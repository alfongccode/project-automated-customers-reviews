```python
from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv(), override=True)
```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from peft import PeftModel

BASE = "cardiffnlp/twitter-roberta-base-sentiment-latest"
LORA = "Bitnick42/roberta-base-review-sentiment-analysis"

tok = AutoTokenizer.from_pretrained(BASE)
model = AutoModelForSequenceClassification.from_pretrained(BASE)
model = PeftModel.from_pretrained(model, LORA)
model = model.merge_and_unload()

model.push_to_hub("Bitnick42/roberta-base-review-sentiment-analysis-latest")
tok.push_to_hub("Bitnick42/roberta-base-review-sentiment-analysis-latest")
```
